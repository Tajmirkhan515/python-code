```python

def divideWholeAddress(fullAddress):
    chunks=fullAddress.split(",")
    
    #extract country, its last position
    countryValue=chunks[-1].split(" ")[-1]
    
    #state, its second last position
    stateValue=chunks[-1].split(" ")[-2]

    #extract zip
    zipValue=chunks[-2].split(" ")[-1]
    #extract of city
    cityValue=chunks[-2].split(" ")[-2]
    
    
    addressValue=chunks[:-2]
    

    return {"address":addressValue , "city":cityValue, "zip":  zipValue, "state":  stateValue, "country":  countryValue}

fullAddress="house# 31, st#12, G-11/3, Islamabad 44000, capital Pakistan"
AddressChunkes=divideWholeAddress(fullAddress)


if AddressChunkes.get('address') and not AddressChunkes.get('state') :
    print("Address: ",AddressChunkes["address"])
    print("City: ",AddressChunkes["city"])
    print("zip: ",AddressChunkes["zip"])
    print("country: ",AddressChunkes["country"])
    
    
if AddressChunkes.get('state') and not AddressChunkes.get('address'):    
    print("City: ",AddressChunkes["city"])
    print("zip: ",AddressChunkes["zip"])
    print("state: ",AddressChunkes["state"])
    print("country: ",AddressChunkes["country"])
    
    
    
if AddressChunkes['address'] and AddressChunkes["state"]:    
    print("City: ",AddressChunkes["city"])
    print("zip: ",AddressChunkes["zip"])    
    print("country: ",AddressChunkes["country"])
```
